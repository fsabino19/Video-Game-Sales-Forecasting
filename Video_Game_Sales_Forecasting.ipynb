```python
# Identify platforms with declining sales

# Get total sales by platform and year
platform_trends = df.groupby(['platform', 'year_of_release']).agg({
    'total_sales': 'sum',
    'name': 'count'  # Number of games
}).reset_index()

platform_trends.columns = ['platform', 'year', 'total_sales', 'game_count']

# Find platforms that had sales in earlier years but zero in recent years
recent_years = [2014, 2015, 2016]
earlier_years = list(range(2000, 2014))  # Adjust range as needed

# Platforms with sales in earlier years
platforms_with_early_sales = df[df['year_of_release'].isin(earlier_years)]['platform'].unique()

# Platforms with sales in recent years
platforms_with_recent_sales = df[df['year_of_release'].isin(recent_years)]['platform'].unique()

# Platforms that used to be popular but now have zero sales
declining_platforms = set(platforms_with_early_sales) - set(platforms_with_recent_sales)
print("Platforms with zero recent sales:", declining_platforms)

# Compare peak sales to recent sales for each platform
platform_decline = []

for platform in df['platform'].unique():
    platform_data = platform_trends[platform_trends['platform'] == platform]
    
    if len(platform_data) > 1:  # Need at least 2 years of data
        peak_sales = platform_data['total_sales'].max()
        recent_sales = platform_data[platform_data['year'] >= 2014]['total_sales'].sum()
        
        if peak_sales > 0:
            decline_percentage = ((peak_sales - recent_sales) / peak_sales) * 100
            
            platform_decline.append({
                'platform': platform,
                'peak_sales': peak_sales,
                'recent_sales': recent_sales,
                'decline_percentage': decline_percentage
            })

decline_df = pd.DataFrame(platform_decline).sort_values('decline_percentage', ascending=False)
```
